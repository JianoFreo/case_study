```python
# Quick interactive plot (plotly)
if 'order_date' in sales.columns and 'total_amount' in sales.columns:
    sales['order_date'] = pd.to_datetime(sales['order_date'], errors='coerce')
    monthly = sales.set_index('order_date').resample('M')['total_amount'].sum().reset_index()
    fig = px.line(monthly, x='order_date', y='total_amount', title='Monthly revenue (sales)')
    fig.show()
else:
    print('order_date or total_amount not present; skipping time-series plot')
```

```python
# Correlation heatmap for numeric columns
num = sales.select_dtypes(include=['number']).dropna(axis=1, how='all')
if not num.empty:
    corr = num.corr()
    plt.figure(figsize=(8,6))
    sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm')
    plt.title('Correlation heatmap (sales)')
    plt.show()
else:
    print('No numeric columns to correlate')
```
