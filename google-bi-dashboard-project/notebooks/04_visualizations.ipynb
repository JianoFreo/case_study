```python
# Scatter: relationship example between numeric cols
num = sales.select_dtypes(include=['number']).columns.tolist()
if len(num) >= 2:
    xcol, ycol = num[0], num[1]
    fig = px.scatter(sales, x=xcol, y=ycol, trendline='ols', title=f'Scatter: {xcol} vs {ycol}')
    fig.show()
else:
    print('Not enough numeric columns for scatter')
```

```python
# Donut / pie: sample breakdown by category (if present)
if 'category' in sales.columns:
    cat = sales.groupby('category')['total_amount'].sum().reset_index()
    fig = px.pie(cat, names='category', values='total_amount', title='Revenue by category', hole=0.4)
    fig.show()
else:
    print('No category column; skipping donut chart')
```

```python
# Line chart: monthly revenue
if 'order_date' in sales.columns and 'total_amount' in sales.columns:
    monthly = sales.set_index('order_date').resample('M')['total_amount'].sum().reset_index()
    fig = px.line(monthly, x='order_date', y='total_amount', title='Monthly revenue')
    fig.update_xaxes(rangeslider_visible=True)
    fig.show()
```

```python
# Bar chart: top products by revenue
if 'product' in sales.columns and 'total_amount' in sales.columns:
    top = sales.groupby('product')['total_amount'].sum().reset_index().sort_values('total_amount', ascending=False).head(10)
    fig = px.bar(top, x='product', y='total_amount', title='Top 10 products by revenue')
    fig.show()
else:
    print('product/total_amount columns not present')
```
