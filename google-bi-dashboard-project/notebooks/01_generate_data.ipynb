```python
# Copy generated files to data/raw for safekeeping
import shutil, os
os.makedirs('data/raw', exist_ok=True)
from pathlib import Path
files = list(Path('data/generated').glob('*.csv'))
for p in files:
    shutil.copy(p, 'data/raw')
print('Copied files to data/raw')
```
